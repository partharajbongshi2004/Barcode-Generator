```python
!pip install python-barcode
import io
import barcode
from barcode.writer import ImageWriter
from IPython.display import display
from PIL import Image

# 1. Define the data or text you want to convert into a barcode
data_to_encode = "COLAB-PROJECT-789"

try:
    # 2. Select Code 128 format (supports both letters and numbers)
    barcode_format = barcode.get_barcode_class("code128")

    # 3. Generate the barcode image directly into an in-memory byte buffer
    buffer = io.BytesIO()
    generated_barcode = barcode_format(data_to_encode, writer=ImageWriter())
    generated_barcode.write(buffer)

    # 4. Open the image from the memory buffer
    buffer.seek(0)
    barcode_image = Image.open(buffer)

    # 5. Display the image directly inside your Google Colab notebook cell
    print(f"📊 Barcode Generated Successfully for: {data_to_encode}\n")
    display(barcode_image)

    # Optional: Save the image file permanently inside your Colab workspace file system
    barcode_image.save("generated_barcode.png")
    print("💾 File saved as 'generated_barcode.png' in your Colab files panel.")

except Exception as e:
    print(f"An error occurred: {e}")


```
